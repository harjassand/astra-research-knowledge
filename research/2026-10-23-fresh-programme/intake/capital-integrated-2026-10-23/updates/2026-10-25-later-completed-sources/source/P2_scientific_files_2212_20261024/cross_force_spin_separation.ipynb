## 1. Exact retained observable and proof

Let the nuclear mechanical Hamiltonian be

\[
H=\sum_i p_i^2/(2m_i)+V(R),\qquad L B=[H,B].
\]

Assume a real scalar potential, canonical coordinates, no velocity-dependent or Berry-force terms, and a time-reversal-invariant thermal mechanical state. For two multiplication operators supported on different nuclei, define

\[
S_{AB}(E)=\sum_{a,n}p_a\langle a|A^\dagger|n\rangle
\langle n|B|a\rangle\delta(E-E_n+E_a),\quad
C_r=\Re\int_{-\infty}^{\infty}E^r S_{AB}(E)\,dE.
\]

In a real eigenbasis the multiplication-operator matrices are symmetric. For odd r this gives

\[
2 C_r=\Re\langle[A^\dagger,L^r B]\rangle.
\]

For i≠j, the first commutator vanishes. In the third commutator all terms confined to coordinate j commute with A_i. The mixed term is

\[
[A_i^\dagger,L^3 B_j]
=\frac{\hbar^4}{m_i m_j}\nabla_i A_i^*\cdot V''_{ij}\cdot\nabla_j B_j.
\]

Hence

\[
C_1=0,\qquad
C_3=\frac{\hbar^4}{2m_i m_j}\Re\langle
\nabla_i A_i^*\cdot V''_{ij}\cdot\nabla_j B_j\rangle.
\]

For plane-wave probes A_i=exp(iq·R_i), B_j=exp(iq·R_j):

\[
C_3(q)=\frac{\hbar^4}{2m_i m_j}\Re\left\langle
e^{iq\cdot(R_j-R_i)}q^T V''_{ij}q\right\rangle.
\]

This is the thermal, phase-weighted mixed curvature of the effective nuclear potential with other nuclear coordinates held fixed in the derivative. Electronic relaxation already included in a Born–Oppenheimer potential remains included. “Direct” does not mean a unique chemical pair potential, an unmediated electronic interaction, or the force obtained after relaxing the other nuclei.

The identity is not restricted to a Gaussian distribution or harmonic vibrations. It requires the *full signed-energy spectrum*, including all orders of nuclear motion. At finite q the result is not simply an unweighted mean spring constant. Dividing by a static structure factor is justified only with additional assumptions: generally <K exp(iqr)> = <K><exp(iqr)> + Cov(K,exp(iqr)). Powder data retain only orientationally averaged projections.

For a centered, finite-variance Gaussian resolution kernel and complete integration, C3_measured=C3+3 sigma_E² C1. This does not control counting noise, energy offsets, truncated integration, long instrumental tails, or contamination by same-atom terms.

For linear coordinate probes in a harmonic model, C_(2n+1)=(hbar^(2n+2)/2)[M^(-1/2)D^n M^(-1/2)]_ij. The n=1 term tests the direct Hessian; n=2 can contain paths through other coordinates. A zero signed cross moment can also arise from cancellation and is not a universal certificate of absent interaction.

## 3. Nuclear scattering: physical labels, not optical masks

For neutron spin Pauli matrices σ and nuclear angular momentum I, write the real bound scattering-length operator

\[
b_s=a_s+c_s\,\boldsymbol\sigma\cdot\mathbf I_s,
\quad a_s=\frac{(I+1)b_++Ib_-}{2I+1},
\quad c_s=\frac{b_+-b_-}{2I+1}.
\]

With unpolarized incident neutrons and final spin unobserved, the instantaneous spin weight is

\[
G_{ij}=a_i a_j+c_i c_j\langle\mathbf I_i\cdot\mathbf I_j\rangle\quad(i\ne j),
\qquad G_{ii}=a_i^2+c_i^2I_i(I_i+1).
\]

The spin weight matrix is positive semidefinite. Mechanical intensities are Tr(G S(q,E)), with S a positive spectral matrix, when the spin and mechanical states factor and spin dynamics are either negligible over the scattering event or explicitly included. Nuclear spin transitions shift energy; equal static self weights alone do not cancel every energy-resolved self line.

Illustrative constants are aH=-3.74 fm, aD=6.67 fm, cH=29.12 fm and cD=2.85 fm. These are approximate low-energy values, not precision or epithermal calibration. The NIST compilation gives the underlying scale, while an actual experiment needs the channel lengths and their energy dependence in its measurement band.

For independent polarized nuclei let v_s=c_s I_s P_s. Independent sign reversals give

\[
I_{++}-I_{+-}-I_{-+}+I_{--}
=8v_Hv_D\,\Re S_{HD}.
\]

The double difference removes self, H–H and D–D contributions before taking a moment. It retains **all H–D pairs**. Averaging the scattering amplitude and then squaring would incorrectly discard the quadratic spin contrast of an unpolarized neutron beam.

Independent H and D RF reversals were demonstrated in the same frozen HD target (Wei et al., PSTP2013). This is evidence for the two controls, not for their combination with high-energy neutron spectroscopy. Four settings cease to be a clean rectangle when reversal losses differ. In the explicit equal-loss 0.9 Gray loop, the nuisance matrix [1,PH²,PD²,PH PD] has rank 3 and the target is not identifiable. A predetermined 12-setting pulse sequence below restores identification, including single-species nuisance columns, with a noise cost.

## 4. Complete executable implementation

The following code is self-contained apart from NumPy and SciPy. Units are eV, Å, amu and fm. The harmonic calculations use one stiffness matrix for modes, displacement statistics, all-order plane-wave spectral moments, and positive raw intensities. The spin calculations use finite-dimensional unitary propagation from a common calibrated spin Hamiltonian. All supplied mechanical springs, geometry and temperatures are visible in the functions.

```python
"""Executable calculations for nuclear-spin contrast and mixed curvature.

This is a conditional measurement-model calculation, not a fitted HD solid.
The final notebook will retain proofs, source scope, this code and its outputs.
Units: energies eV, lengths Angstrom, nuclear masses amu, scattering lengths fm.
"""
import numpy as np
from scipy import constants
from scipy.optimize import brentq
from scipy.linalg import null_space, expm
from scipy.stats import poisson
from math import comb, factorial
import json

CM = constants.hbar**2/(2*constants.atomic_mass*constants.electron_volt*1e-20)
CN = constants.hbar**2/(2*constants.neutron_mass*constants.electron_volt*1e-20)
KB = constants.Boltzmann/constants.electron_volt

def angular_momentum(I):
    m = np.arange(I, -I-0.1, -1)
    zp = np.diag(m)
    plus = np.zeros((len(m),len(m)),complex)
    for a, ma in enumerate(m):
        if a > 0:
            plus[a-1,a] = np.sqrt(I*(I+1)-ma*(ma+1))
    return [(plus+plus.T)/2, (plus-plus.T)/(2j), zp]

def spin_density(I, P):
    m = np.arange(I,-I-0.1,-1)
    def mean(lam):
        w=np.exp(lam*m-np.max(lam*m)); w/=w.sum()
        return w@m/I
    lam=brentq(lambda t:mean(t)-P,-80,80)
    w=np.exp(lam*m-np.max(lam*m)); w/=w.sum()
    return np.diag(w)

def spin_check(PH=.6, PD=.18, aH=-3.74, aD=6.67, gH=14.56, gD=2.85):
    # g_s = c_s I_s; operator b_s = a_s + c_s sigma . I_s.
    sx=2*np.array(angular_momentum(.5))
    ih, idd=angular_momentum(.5), angular_momentum(1.)
    ident=np.eye(12)
    bh=aH*ident+sum(2*gH*np.kron(np.kron(sx[k],ih[k]),np.eye(3)) for k in range(3))
    bd=aD*ident+sum(gD*np.kron(np.kron(sx[k],np.eye(2)),idd[k]) for k in range(3))
    out={}
    for sh,sd in [(1,1),(1,-1),(-1,1),(-1,-1)]:
        rho=np.kron(np.kron(np.eye(2)/2,spin_density(.5,sh*PH)),spin_density(1,sd*PD))
        vals=np.array([np.trace(rho@bh@bh),np.trace(rho@bd@bd),np.trace(rho@bh@bd)]).real
        exact=np.array([aH*aH+3*gH*gH,aD*aD+2*gD*gD,aH*aD+sh*sd*gH*gD*PH*PD])
        assert np.allclose(vals,exact,atol=1e-10)
        out[f'{sh:+d},{sd:+d}']=vals.tolist()
    out['four_state_cross_coefficient_fm2']=8*gH*gD*PH*PD
    return out

def model():
    # Two HD-like collinear units in harmonic cages. Molecular stiffness from
    # the approximate 3813 cm^-1 harmonic HD scale. Other constants specified,
    # not sourced or claimed to describe actual solid HD.
    m=np.array([1.007825,2.014102,1.007825,2.014102])
    r=np.array([0.,.7414,3.8,4.5414])
    pairs=[(0,1,35.7),(2,3,35.7),(1,2,.08),(0,3,.02),
           (0,2,.03),(1,3,.03)]
    K=np.eye(4)*.05
    for i,j,k in pairs:
        e=np.eye(4)[i]-np.eye(4)[j]
        K+=k*np.outer(e,e)
    evals,V=np.linalg.eigh(2*CM*K/np.sqrt(m[:,None]*m[None,:]))
    E=np.sqrt(evals)
    # Columns of L are zero-point displacement amplitudes.
    L=V*np.sqrt(CM/(m[:,None]*E[None,:]))
    U=L@L.T
    return m,r,K,E,L,U

def bell(cumulants, nmax):
    b=np.zeros(nmax+1,dtype=np.result_type(cumulants,float)); b[0]=1
    for n in range(1,nmax+1):
        b[n]=sum(comb(n-1,k-1)*cumulants[k-1]*b[n-k] for k in range(1,n+1))
    return b

def harmonic_moments(q, r, E, L, nmax=6,temperature=0.):
    occupation=np.zeros_like(E) if temperature==0 else 1/np.expm1(E/(KB*temperature))
    U=(L*(2*occupation+1))@L.T
    out=np.zeros((nmax+1,len(r),len(r)),complex)
    for i in range(len(r)):
        for j in range(len(r)):
            alpha=q*q*L[i]*L[j]
            c0=np.exp(1j*q*(r[j]-r[i])-.5*q*q*(U[i,i]+U[j,j])+alpha@(2*occupation+1))
            cumul=np.array([alpha@(E**k*(1+(1+(-1)**k)*occupation)) for k in range(1,nmax+1)])
            out[:,i,j]=c0*bell(cumul,nmax)
    return out

def scattering_matrix(sh,sd,PH=.6,PD=.18):
    a=np.array([-3.74,6.67,-3.74,6.67])
    g=np.array([14.56,2.85,14.56,2.85])
    spin=np.array([.5,1,.5,1])
    pol=np.array([sh*PH,sd*PD,sh*PH,sd*PD])
    G=np.outer(a,a)+np.outer(g*pol,g*pol)
    G[np.diag_indices(4)]=a*a+g*g*(spin+1)/spin
    assert np.linalg.eigvalsh(G).min()>0
    return G

def shell_design(qs,m,r,U):
    # columns: common intramolecular k, near intermolecular k, far k.
    groups=[[(0,1),(2,3)],[(1,2)],[(0,3)]]
    F=np.zeros((len(qs),len(groups)))
    for a,q in enumerate(qs):
        for b,grp in enumerate(groups):
            for i,j in grp:
                var=U[i,i]+U[j,j]-2*U[i,j]
                F[a,b]+=-2*CM**2*q*q/(m[i]*m[j])*np.cos(q*(r[j]-r[i]))*np.exp(-q*q*var/2)
    return F

def recovery(qs=np.linspace(4,12,81), Ncounts=1e9):
    m,r,K,E,L,U=model()
    cross=[]; m0=[]; m6=[]; max_null=0.; max_diff=0.
    D=8*14.56*2.85*.6*.18
    signs=[(1,1),(1,-1),(-1,1),(-1,-1)]
    for q in qs:
        M=harmonic_moments(q,r,E,L)
        mh=M[:,[0,0,2,2],[1,3,1,3]].sum(axis=1).real
        ms=np.array([np.einsum('ij,nij->n',scattering_matrix(sh,sd),M).real for sh,sd in signs])
        dd=(ms[0]-ms[1]-ms[2]+ms[3])/D
        max_null=max(max_null,abs(mh[1]))
        max_diff=max(max_diff,np.max(abs(dd-mh)))
        cross.append(mh[3]);m0.append(ms[:,0].mean());m6.append(ms[:,6].mean())
    cross,m0,m6=map(np.array,(cross,m0,m6))
    F=shell_design(qs,m,r,U)
    assert np.allclose(cross,F@np.array([35.7,.08,.02]),atol=1e-15)
    # Same incident exposure per q and state; Ncounts includes elastic counts.
    exposure=Ncounts/(4*m0.sum())
    variances=4*m6/(exposure*D**2)
    Fw=F/np.sqrt(variances[:,None])
    cov=np.linalg.inv(Fw.T@Fw)
    rec=np.linalg.solve(Fw.T@Fw,Fw.T@(cross/np.sqrt(variances)))
    W=cov@(F.T/variances)
    normed=F/np.linalg.norm(F,axis=0)
    # A common displacement of every molecular bond length is not a fit to a
    # force. This quantifies independent structural calibration sensitivity.
    rp=r.copy();rp[[1,3]]+=1e-3
    Fb=shell_design(qs,m,rp,U)
    bias=W@(Fb@np.array([35.7,.08,.02])-cross)
    gain_tolerance=[]
    M3even=[]
    for q in qs:
        M=harmonic_moments(q,r,E,L)
        G=sum(scattering_matrix(sh,sd) for sh,sd in signs)/4
        M3even.append(np.einsum('ij,ij->',G,M[3]).real)
    M3even=np.array(M3even)
    # Correlated raw gain modulation eta*sh*sd produces spurious dd=4eta Ieven.
    gain_bias_per_unit=W@(4*M3even/D)
    return dict(mode_energies_eV=E.tolist(),pair_variances_A2={
        f'{i}-{j}':float(U[i,i]+U[j,j]-2*U[i,j]) for i,j in [(0,1),(2,3),(1,2),(0,3)]},
        max_C1_abs=float(max_null),max_spin_projection_error=float(max_diff),
        recovered_k=rec.tolist(),sigma_at_Ncounts=np.sqrt(np.diag(cov)).tolist(),
        Ncounts=Ncounts,raw_design_condition=float(np.linalg.cond(F)),
        column_normalized_condition=float(np.linalg.cond(normed)),
        noise_weighted_condition=float(np.linalg.cond(Fw)),
        bias_from_1mA_bond_offset=bias.tolist(),
        bias_per_unit_double_odd_gain=gain_bias_per_unit.tolist(),
        N_for_10pct_near=float(Ncounts*(np.sqrt(cov[1,1])/.008)**2)),(qs,cross,F,W,variances)

def identifiability():
    E=np.array([.02,.04,.065,.10,.15,.22])
    one=np.ones(6); lam=E**2
    fs=[1/(np.tanh(E/(2*KB*T))*E) for T in [300,500,900]]
    def perp(x,cols):
        A=np.array(cols).T
        Q,_=np.linalg.qr(A/np.linalg.norm(A,axis=0))
        return x-Q@(Q.T@x)
    base=perp(fs[0],[one,lam]);base*=.06/np.max(abs(base))
    v=perp(lam,[one]+fs);v/=np.max(abs(v))
    cc=[base,base+.05*v]
    Ks=[];Us=[]
    for c in cc:
        blocks=[]
        for cv in c:
            w,V=np.linalg.eigh(np.array([[1/6,cv],[cv,1/6]]))
            blocks.append((V*np.sqrt(w))@V.T)
        O2=np.hstack(blocks)
        O=np.vstack([O2,null_space(O2).T])
        assert np.allclose(O@O.T,np.eye(12),atol=1e-12)
        K=12/(2*CM)*(O*np.repeat(E**2,2))@O.T
        Ks.append(K)
        Us.append([CM/12*(O2*np.repeat(f,2))@O2.T for f in fs])
    return dict(cA=cc[0].tolist(),cB=cc[1].tolist(),
        K12_eV_A2=[k[0,1] for k in Ks],
        max_observed_covariance_difference=float(np.max(abs(np.array(Us[0])-np.array(Us[1])))),
        minimum_K_eigenvalue=min(np.linalg.eigvalsh(k).min() for k in Ks))

def transitions(q, r, E, L, tail=1e-14):
    """Ground-state harmonic transitions, all phonon orders to a checked cutoff.

    Each state is generated by the SAME normal modes and plane-wave operator.
    No independent line intensities, self background, or cross lines supplied.
    """
    lam=q*q*L*L
    ns=np.array([int(poisson.ppf(1-tail,np.max(lam[:,j]))+1) for j in range(len(E))])
    indexes=np.indices(tuple(ns)).reshape(len(E),-1).T
    energies=indexes@E
    amp=np.zeros((len(energies),len(r)),complex)
    for i in range(len(r)):
        prod=np.ones(len(energies))*np.exp(-.5*lam[i].sum())
        for j in range(len(E)):
            coeff=[1.]
            for n in range(1,ns[j]):
                coeff.append(coeff[-1]*q*L[i,j]/np.sqrt(n))
            prod*=np.array(coeff)[indexes[:,j]]
        amp[:,i]=np.exp(1j*q*r[i])*prod
    return energies,amp

def finite_window_tests(qs=np.linspace(4,12,81)):
    m,r,K,E,L,U=model()
    result,(_,full,F,W,var)=recovery(qs)
    policies=['0.6 eV','1 eV','2 eV','Ei=2 eV kinematic']
    outputs={s:[] for s in policies}
    bounds=[];rows=[];worst_m6_error=0.
    for q in qs:
        e,A=transitions(q,r,E,L)
        AH=A[:,0]+A[:,2];AD=A[:,1]+A[:,3]
        cross=(AH.conj()*AD).real
        ideal=harmonic_moments(q,r,E,L)
        numerical6=A.conj().T@((e**6)[:,None]*A)
        worst_m6_error=max(worst_m6_error,np.max(abs(numerical6-ideal[6]))/np.max(abs(ideal[6])))
        cuts=[.6,1.,2.,2*np.sqrt(CN*2.)*q-CN*q*q]
        for s,cut in zip(policies,cuts):
            mask=e<=cut
            outputs[s].append(float(cross[mask]@(e[mask]**3)))
        # Independently measurable positive spectral-tail certificate.
        t=e>1.
        ta=float((abs(AH[t])**2)@(e[t]**3))
        tb=float((abs(AD[t])**2)@(e[t]**3))
        b=np.sqrt(ta*tb);bounds.append(b)
        if np.any(np.isclose(q,[4,6,8,10,12])):
            rows.append(dict(q=float(q),states=len(e),C3_full=float(ideal[3][np.ix_([0,2],[1,3])].sum().real),
                C3_below_1eV=outputs['1 eV'][-1],tail_bound_1eV=float(b),
                Emax_at_Ei2=float(cuts[-1])))
    fit={s:(W@np.array(vals)).tolist() for s,vals in outputs.items()}
    # Non-circular IF these are acquired auto-spectral tails, not just simulated.
    propagated_tail=np.abs(W)@np.array(bounds)
    return dict(rows=rows,fitted_k_after_truncation=fit,
        propagated_1eV_tail_bound=propagated_tail.tolist(),
        max_relative_full_M6_enumeration_error=float(worst_m6_error))

def pulse_path(sequence, etaH=.9, etaD=.9):
    """Each RF pulse reverses one species with specified retained magnitude.
    Efficiencies here are stress-test inputs; actual P is read by NMR.
    """
    x,y=1.,1.;states=[(x,y)]
    for s in sequence:
        if s=='H': x*=-etaH
        elif s=='D': y*=-etaD
        else: raise ValueError(s)
        states.append((x,y))
    return np.array(states)

def contrast_design(states, linear=False):
    x,y=states.T
    cols=[np.ones(len(x)),x*x,y*y]
    if linear:cols.extend([x,y])
    cols.append(x*y)
    X=np.array(cols).T
    target=np.eye(X.shape[1])[-1]
    w=np.linalg.lstsq(X.T,target,rcond=1e-12)[0]
    return dict(states=states.tolist(),rank=int(np.linalg.matrix_rank(X)),
        ncolumns=X.shape[1],target_residual=float(np.linalg.norm(X.T@w-target)),
        scaled_design_condition=float(np.linalg.cond(X/np.linalg.norm(X,axis=0))),
        equal_count_noise_multiplier=float(np.sqrt(len(w)*(w@w))),
        weights=w.tolist())

def physical_contrast_tests():
    # A lossy four-state Gray loop can be NONidentifying, although an ideal
    # lossless four-state rectangle isolates the cross coefficient perfectly.
    bad=contrast_design(pulse_path('HDH'))
    # Predetermined sequence; no pulse optimization or assumption of reset.
    goodseq='HHDHDDHDHHD'
    good=contrast_design(pulse_path(goodseq),linear=True)
    return dict(equal_loss_four_states=bad,
        twelve_states_with_single_species_nuisances=good,
        ideal_noise_multiplier=1.,sequence=goodseq)

def pair_spin_operators():
    H=[np.kron(x,np.eye(3)) for x in angular_momentum(.5)]
    D=[np.kron(np.eye(2),x) for x in angular_momentum(1.)]
    return H,D

def prepare_pair_order(P=.6, J=47.2, detuning=0.):
    """Explicit on-proton pulses and intramolecular scalar-J evolution.

    Deuterium is initially maximally mixed, an explicit initialization
    requirement. detuning is a proton frequency offset, in Hz.
    No deuteron pulse or hypothetical independent mechanical port is used.
    """
    H,D=pair_spin_operators(); Ix,Iy,Iz=H; Sx,Sy,Sz=D
    rho=np.eye(6)/6+P*Iz/3
    U1=expm(-1j*np.pi/2*Iy)
    tj=1/(4*J)
    Uj=expm(-2j*np.pi*tj*(J*Iz@Sz+detuning*Iz))
    U2=expm(-1j*np.pi/2*Ix)
    U=U2@Uj@U1
    plus=U@rho@U.conj().T
    flip=expm(-1j*np.pi*Ix)
    minus=flip@plus@flip.conj().T
    return plus,minus

def partial_pair(rho):
    t=rho.reshape(2,3,2,3)
    return np.einsum('abcb->ac',t),np.einsum('abad->bd',t)

def two_molecule_spin_matrix(rho):
    H,D=pair_spin_operators()
    ops=[[np.kron(x,np.eye(6)) for x in H],
         [np.kron(x,np.eye(6)) for x in D],
         [np.kron(np.eye(6),x) for x in H],
         [np.kron(np.eye(6),x) for x in D]]
    rr=np.kron(rho,rho)
    a=np.array([-3.74,6.67,-3.74,6.67])
    c=np.array([29.12,2.85,29.12,2.85])
    G=np.outer(a,a).astype(complex)
    for i in range(4):
        for j in range(4):
            G[i,j]+=c[i]*c[j]*sum(np.trace(rr@ops[i][k]@ops[j][k]) for k in range(3))
    assert np.linalg.eigvalsh(G).min()>0
    return G.real

def pair_order_test(P=.6):
    H,D=pair_spin_operators();p,m=prepare_pair_order(P)
    ph,pd=partial_pair(p);mh,md=partial_pair(m)
    Gp,Gm=two_molecule_spin_matrix(p),two_molecule_spin_matrix(m)
    delta=Gp-Gm
    target=np.zeros((4,4))
    coeff=2*P/3*29.12*2.85
    for i,j in [(0,1),(2,3)]: target[i,j]=target[j,i]=coeff
    assert np.allclose(delta,target,atol=1e-11)
    # A GLOBAL phase cycle removes single-molecule transverse terms but does
    # NOT produce independent dephasing of the many-molecule density matrix.
    # For these two-point scattering weights the residual H-H correlations
    # are identical in both sign settings and cancel.
    R=expm(-1j*np.pi*H[2])
    pr=R@p@R.conj().T;mr=R@m@R.conj().T
    phase_delta=(Gp+two_molecule_spin_matrix(pr)-Gm-two_molecule_spin_matrix(mr))/2
    return dict(P=P,gate_time_s=1/(4*47.2),
        minimum_density_eigenvalue=float(min(np.linalg.eigvalsh(p).min(),np.linalg.eigvalsh(m).min())),
        H_marginal_difference=float(np.max(abs(ph-mh))),
        D_marginal_difference=float(np.max(abs(pd-md))),
        pair_dot_correlations=[float(np.trace(rr@sum(H[k]@D[k] for k in range(3))).real) for rr in [p,m]],
        difference_scattering_matrix_fm2=delta.tolist(),
        desired_spectrum_multiplier_fm2=2*coeff,
        global_phase_cycle_mask_error=float(np.max(abs(phase_delta-target))))

def pair_spin_energy_moments(P=.6,B=1.4,J=47.2,nmax=6):
    """Exact local spin transition moments, including Zeeman and scalar J.

    Both states are diagonal pair order after a local marginal phase average.
    For scattering from one molecule this marginal is also produced by a
    global phase cycle. Neutron spin is unpolarized and final spin unobserved.
    H_s is separable from mechanics; J(R) corrections are NOT modeled here.
    """
    H,D=pair_spin_operators(); hz=constants.h/constants.electron_volt
    hs=hz*(-42.57747892e6*B*H[2]-6.5359023e6*B*D[2]+J*H[2]@D[2])
    en=np.repeat(np.diag(hs).real,2)
    # Ordering is nuclear(6) x neutron(2).
    pauli=[2*x for x in angular_momentum(.5)]
    bs=[]
    for a,c,ops in [(-3.74,29.12,H),(6.67,2.85,D)]:
        bs.append(a*np.eye(12)+c*sum(np.kron(ops[k],pauli[k]) for k in range(3)))
    deltaE=en[:,None]-en[None,:] # final-row minus initial-column
    result=[]
    for sign in [1,-1]:
        rho=np.eye(6)/6+sign*P/3*(H[2]@D[2])
        population=np.repeat(np.diag(rho).real/2,2)
        M=np.zeros((nmax+1,2,2),complex)
        for r in range(nmax+1):
            weight=(deltaE**r)*population[None,:]
            for i in range(2):
                for j in range(2):
                    M[r,i,j]=np.sum(weight*bs[i].conj()*bs[j])
        result.append(M.real)
    return np.array(result)

def pair_tag_recovery(qs=np.linspace(4,12,81),P=.6,Ncounts=1e9):
    """One-parameter illustrative intrabond recovery, not the main observable.

    The two molecular spring constants are assumed equal. Marginal-dephased
    pair order eliminates ALL inter-molecular contributions in the mean.
    Its two raw spectra retain self/host shot noise; host is omitted here,
    making these uncertainties optimistic lower benchmarks for a dilute HD
    matrix. Exact moments of spin + SAME mechanical Hamiltonian are used.
    """
    m,r,K,E,L,U=model()
    sm=pair_spin_energy_moments(P,nmax=6)
    Dcoef=4*P*29.12*2.85/3
    a=np.array([-3.74,6.67,-3.74,6.67])
    raw=[];sig=[];pred=[];spin_leak=[]
    for q in qs:
        M=harmonic_moments(q,r,E,L)
        vals=np.zeros((2,7))
        for state in range(2):
            # Inter-molecule spin averages vanish. Spin-independent coherent
            # amplitude survives, but is identical between the two settings.
            for n in range(7):
                G0=np.outer(a,a).copy()
                for i,j in [(0,1),(2,3)]:
                    inds=np.ix_([i,j],[i,j])
                    G0[inds]=sm[state,0]
                vals[state,n]=np.einsum('ij,ij->',G0,M[n]).real
                for k in range(1,n+1):
                    for i,j in [(0,1),(2,3)]:
                        vals[state,n]+=comb(n,k)*np.einsum('ij,ij->',sm[state,k],M[n-k][np.ix_([i,j],[i,j])]).real
        pure=float((M[3,0,1]+M[3,2,3]).real)
        pred.append(pure);sig.append((vals[0,3]-vals[1,3])/Dcoef)
        spin_leak.append(sig[-1]-pure);raw.append(vals)
    raw=np.array(raw);pred=np.array(pred);sig=np.array(sig)
    F=shell_design(qs,m,r,U)[:,0]
    exposure=Ncounts/raw[:,:,0].sum()
    variance=raw[:,:,6].sum(axis=1)/(exposure*Dcoef**2)
    weight=(F/variance)/(F@(F/variance))
    khat=float(weight@sig);sigma=float(np.sqrt(1/(F@(F/variance))))
    # Infinite-window cubic moment is the primary observable. This k fit
    # illustrates noise only and grants static structural calibration.
    return dict(P=P,signal_multiplier_fm2=Dcoef,
        k_from_full_moment=khat,k_true=35.7,
        maximum_spin_energy_correction_eV3=float(np.max(np.abs(spin_leak))),
        k_bias_from_known_spin_energies=khat-35.7,
        sigma_k_at_Ncounts=sigma,Ncounts=Ncounts,
        N_for_10pct_k=float(Ncounts*(sigma/3.57)**2))

def pair_control_diagnostics():
    Pthermal=float(np.tanh(constants.h*42.57747892e6*1.4/(2*constants.k*4)))
    rows=[]
    for df in [0.,1.,5.,10.,20.]:
        p,m=prepare_pair_order(.6,detuning=df)
        H,D=pair_spin_operators()
        c=float(np.trace((p-m)@sum(H[k]@D[k] for k in range(3))).real)
        ph,pd=partial_pair(p);mh,md=partial_pair(m)
        rows.append(dict(detuning_Hz=df,pair_tag_fraction=c/.4,
            H_marginal_difference=float(np.max(abs(ph-mh))),
            D_marginal_difference=float(np.max(abs(pd-md)))))
    return dict(thermal_P_H_4K_1p4T=Pthermal,
        optimistic_exponential_gate_survival_at_T2star=[
            [t,float(np.exp(-1/(4*47.2*t)))] for t in [.02,.3]],
        detuning_rows=rows,
        high_P_benchmark=pair_tag_recovery(P=.6),
        thermal_P_benchmark=pair_tag_recovery(P=Pthermal))

def hh_operators():
    s=angular_momentum(.5)
    return [np.kron(x,np.eye(2)) for x in s],[np.kron(np.eye(2),x) for x in s]

def hh_scattering_two_molecules(rho):
    H,D=hh_operators(); ops=[]
    for group in [H,D]:ops.append([np.kron(x,np.eye(4)) for x in group])
    for group in [H,D]:ops.append([np.kron(np.eye(4),x) for x in group])
    rr=np.kron(rho,rho);G=np.ones((4,4))*(-3.74)**2
    for i in range(4):
        for j in range(4):
            G[i,j]+=29.12**2*sum(np.trace(rr@ops[i][k]@ops[j][k]).real for k in range(3))
    return G

def raw_reference_bound(residual,Gref):
    """For any PSD mechanical spectral matrix S, |tr(residual S)| <=
    eta tr(Gref S). Gref must be positive definite, independently calibrated.
    This bound does NOT assume a force field or small first moment.
    """
    eig,V=np.linalg.eigh(Gref)
    assert eig.min()>0
    invroot=(V/np.sqrt(eig))@V.T.conj()
    eta=np.max(abs(np.linalg.eigvalsh(invroot@residual@invroot)))
    return float(eta)

def phip_control_test():
    """Published Ru dihydride spin parameters; no Ru mechanical force model.

    Source populations, rather than an ideal Werner approximation, set the
    longitudinal order. A proposed rectangular selective pulse is simulated
    including the acquired scalar coupling. Its B1 is a design value, not a
    demonstrated setting in the source experiment.
    """
    H,D=hh_operators();basis=np.eye(4)
    singlet=(basis[1]-basis[2])/np.sqrt(2)
    trip0=(basis[1]+basis[2])/np.sqrt(2)
    rho=.937*np.outer(singlet,singlet)+.045*np.outer(trip0,trip0)
    rho+=.009*np.outer(basis[0],basis[0])+.009*np.outer(basis[3],basis[3])
    Z=expm(-1j*np.pi*H[2]);X=expm(-1j*np.pi*H[0])
    rm=(rho+Z@rho@Z.conj().T)/2
    rp=X@rm@X.conj().T
    alpha=.937+.045-.009-.009
    assert np.allclose(rm,np.eye(4)/4-alpha*H[2]@D[2])
    assert np.allclose(rp,np.eye(4)/4+alpha*H[2]@D[2])
    gm=hh_scattering_two_molecules(rm);gp=hh_scattering_two_molecules(rp)
    ideal=gp-gm
    wanted=np.zeros((4,4))
    for i,j in [(0,1),(2,3)]:wanted[i,j]=wanted[j,i]=alpha*29.12**2/2
    assert np.allclose(ideal,wanted,atol=1e-10)
    # A rectangular pulse: on-resonance first H rotates pi, detuned second
    # H rotates 2pi. Both see the SAME RF field. Selectivity is physical.
    delta=492.;J=4.6;rf=delta/np.sqrt(3);duration=1/(2*rf)
    hrf=-delta*D[2]+rf*(H[0]+D[0])+J*sum(H[k]@D[k] for k in range(3))
    u=expm(-2j*np.pi*duration*hrf)
    actual=u@rm@u.conj().T
    ga=hh_scattering_two_molecules(actual)
    residual=ga-gm-wanted;gref=(ga+gm)/2
    means=[[float(np.trace(rr@op).real) for op in H+D] for rr in [rm,rp,actual]]
    # A separate counterexample: pair population/two-spin longitudinal order
    # alone cannot certify absence of correlations between molecules.
    # |up down up down> / |down up down up> equally mixed has the same two
    # anticorrelated pair marginals as independent alpha=1 pairs, but large
    # undesired cross-pair correlations.
    b16=np.eye(16);global_classical=(np.outer(b16[5],b16[5])+np.outer(b16[10],b16[10]))/2
    allops=[np.kron(H[2],np.eye(4)),np.kron(D[2],np.eye(4)),
            np.kron(np.eye(4),H[2]),np.kron(np.eye(4),D[2])]
    wrong=float(np.trace(global_classical@allops[0]@allops[3]).real)
    return dict(source_spin_populations=dict(S=.937,T0=.045,Tplus=.009,Tminus=.009),
        longitudinal_alpha=alpha,contrast_fm2=alpha*29.12**2,
        ideal_spin_weight_error=float(np.max(abs(ideal-wanted))),
        selective_pulse=dict(delta_Hz=delta,J_Hz=J,rf_Hz=rf,duration_s=duration,
            B1_uT=rf/42.57747892e6*1e6,
            state_trace_distance=float(np.sum(abs(np.linalg.eigvalsh(actual-rp)))/2),
            six_one_spin_means_before_ideal_after_actual=means,
            raw_reference_leakage_eta=raw_reference_bound(residual,gref),
            maximum_unwanted_pair_weight_fm2=float(np.max(abs(residual[np.ix_([0,1],[2,3])]))),
            target_pair_weight_error_fm2=float(residual[0,1])),
        same_pair_marginals_counterexample_wrong_inter_molecular_zz=wrong)

def phip_physical_family(nsettings=16,hold_s=.02,hold_steps=None):
    """Same RF amplitude, varied pulse duration, followed by a timed hold.

    This supplies a physical contrast family instead of freely specified
    scattering masks. Its controls use source delta/J. Pair correlations and
    one-spin means are propagated together. All molecule pairs start in a
    product of IDENTICAL intramolecular states; intermolecular spin transfer,
    relaxation, ligands and nonuniform RF must be independently checked.

    Intermolecular weights average products of instantaneous means, NOT
    products of averaged means. A common Larmor frame is valid for scalar
    spin dot-products; it is not used to discard energy shifts in moments.
    """
    H,D=hh_operators();alpha=.964;rm=np.eye(4)/4-alpha*H[2]@D[2]
    delta=492.;J=4.6;rf=delta/np.sqrt(3)
    hrf=-delta*D[2]+rf*(H[0]+D[0])+J*sum(H[k]@D[k] for k in range(3))
    hfree=delta/2*(H[2]-D[2])+J*sum(H[k]@D[k] for k in range(3))
    ev,V=np.linalg.eigh(hfree)
    frequencies=(ev[:,None]-ev[None,:]).reshape(-1)
    def phase_average(f):
        return np.exp(-1j*np.pi*f*hold_s)*np.sinc(f*hold_s)
    av1=phase_average(frequencies)
    av2=phase_average(frequencies[:,None]+frequencies[None,:])
    oe=[V.conj().T@o@V for o in H+D+[sum(H[k]@D[k] for k in range(3))]]
    if hold_steps is not None:
        times=np.linspace(0,hold_s,hold_steps)
        holdw=np.ones(hold_steps);holdw[[0,-1]]=.5;holdw/=holdw.sum()
        Us=np.array([(V*np.exp(-2j*np.pi*ev*t))@V.conj().T for t in times])
    durations=np.linspace(0,4/(2*rf),nsettings)
    rows=[]
    for t in durations:
        u=expm(-2j*np.pi*t*hrf);rho=u@rm@u.conj().T
        if hold_steps is None:
            re=V.conj().T@rho@V
            coeff=np.array([(re*o.T).reshape(-1) for o in oe])
            av=coeff@av1
            aa=sum(coeff[k]@av2@coeff[k] for k in range(3))
            bb=sum(coeff[k+3]@av2@coeff[k+3] for k in range(3))
            ab=sum(coeff[k]@av2@coeff[k+3] for k in range(3))
            rows.append(np.array([1.,aa,bb,ab,av[2],av[5],av[6]]).real)
        else:
            rr=Us@rho@Us.conj().transpose(0,2,1)
            a=np.array([np.einsum('tij,ji->t',rr,o).real for o in H]).T
            b=np.array([np.einsum('tij,ji->t',rr,o).real for o in D]).T
            qq=np.einsum('tij,ji->t',rr,sum(H[k]@D[k] for k in range(3))).real
            rows.append([1.,holdw@(a*a).sum(axis=1),holdw@(b*b).sum(axis=1),
                holdw@(a*b).sum(axis=1),holdw@a[:,2],holdw@b[:,2],holdw@qq])
    A=np.array(rows)
    scales=np.linalg.norm(A,axis=0);An=A/scales
    target=np.zeros(7);target[-1]=alpha/2
    w=np.linalg.lstsq(An.T,target/scales,rcond=1e-10)[0]
    residual=A.T@w-target
    # Sum weights removes all unchanged species. The next three constraints
    # remove all inter-molecular AA, BB and AB pairs, irrespective of their
    # mechanical correlations or separation, under the product-spin premise.
    # z columns allow a thermal polarized spectator or residual beam effects.
    return dict(hold_s=hold_s,nsettings=nsettings,
        time_averaging='analytic finite-window' if hold_steps is None else 'trapezoidal diagnostic',
        pulse_durations_ms=(durations*1000).tolist(),rf_Hz=rf,
        calibration_columns=['1','<mA.mA>','<mB.mB>','<mA.mB>','<mAz>','<mBz>','<IA.IB>'],
        calibration_matrix=A.tolist(),weights=w.tolist(),
        rank=int(np.linalg.matrix_rank(An,tol=1e-10)),
        normalized_singular_values=np.linalg.svd(An,compute_uv=False).tolist(),
        maximum_projection_residual=float(np.max(abs(residual))),
        equal_intensity_count_SD_relative_to_ideal_two_states=float(np.sqrt(nsettings*(w@w))/2),
        worst_case_absolute_mean_calibration_amplification=float(np.abs(w).sum()))

def common_heavy_bridge(kappa=0.,temperature=295.):
    """Controlled two H--M--H model, NOT a Ru force-field prediction.

    A spinless scaffold of 100 amu supports two light sites. All intra- and
    interunit mechanical correlations, force derivatives, spectral moments
    and positive counting intensities come from this ONE quadratic energy.
    The actual Ru experiment supplies only spin-control parameters elsewhere.
    """
    m=np.array([1.007825,100.,1.007825,1.007825,100.,1.007825])
    r=np.array([-1.6,0.,1.6,3.4,5.,6.6]);K=np.eye(6)*.05
    pairs=[(0,1,12.),(1,2,12.),(3,4,12.),(4,5,12.),
           (0,2,kappa),(3,5,kappa),(2,3,.08),(0,5,.03),
           (0,3,.04),(2,5,.05),(1,4,.2)]
    for i,j,k in pairs:
        v=np.eye(6)[i]-np.eye(6)[j];K+=k*np.outer(v,v)
    ev,V=np.linalg.eigh(2*CM*K/np.sqrt(m[:,None]*m[None,:]));E=np.sqrt(ev)
    L=V*np.sqrt(CM/(m[:,None]*E[None,:]))
    nth=1/np.expm1(E/(KB*temperature));U=(L*(2*nth+1))@L.T
    return m,r,K,E,L,U

def bridge_scattering_matrix(row):
    a=np.array([-3.74,7.,-3.74,-3.74,7.,-3.74]);c=29.12
    G=np.outer(a,a)
    for i in [0,2,3,5]:G[i,i]=(-3.74)**2+3*14.56**2
    for i,j in [(0,2),(3,5)]:G[i,j]+=c*c*row[-1];G[j,i]=G[i,j]
    for i,j,col in [(0,3,1),(2,5,2),(0,5,3),(2,3,3)]:
        G[i,j]+=c*c*row[col];G[j,i]=G[i,j]
    # Two spinless scaffold atoms contribute a rank-one coherent block;
    # positivity, not strict positive definiteness, is the physical condition.
    assert np.linalg.eigvalsh(G).min()>-1e-10
    return G

def bridge_discriminator(kappa=0.,Ncounts=1e9):
    cal=phip_physical_family();A=np.array(cal['calibration_matrix']);w=np.array(cal['weights'])
    m,r,K,E,L,U=common_heavy_bridge(kappa);qs=np.linspace(1,8,71)
    gs=np.array([bridge_scattering_matrix(row) for row in A]);Dcoef=.964*29.12**2
    raw=[];pred=[];null=[];F=[];fifth=[]
    for q in qs:
        M=harmonic_moments(q,r,E,L,temperature=295.)
        raw.append(np.einsum('kij,nij->kn',gs,M).real)
        pred.append(float((M[3,0,2]+M[3,3,5]).real))
        null.append(float((M[1,0,2]+M[1,3,5]).real))
        fifth.append(float((M[5,0,2]+M[5,3,5]).real))
        f=0.
        for i,j in [(0,2),(3,5)]:
            vv=U[i,i]+U[j,j]-2*U[i,j]
            f+=-2*CM**2*q*q/(m[i]*m[j])*np.cos(q*(r[j]-r[i]))*np.exp(-q*q*vv/2)
        F.append(f)
    raw=np.array(raw);pred=np.array(pred);F=np.array(F)
    signal=raw[:,:,3]@w/Dcoef
    exposure=Ncounts/raw[:,:,0].sum()
    variance=(raw[:,:,6]@(w*w))/(exposure*Dcoef**2)
    W=(F/variance)/(F@(F/variance))
    # Force after allowing the environment coordinates to relax: Schur
    # complement, distinct from the fixed-environment bare mixed derivative.
    obs=[0,2];env=[1,3,4,5]
    Keff=K[np.ix_(obs,obs)]-K[np.ix_(obs,env)]@np.linalg.solve(K[np.ix_(env,env)],K[np.ix_(env,obs)])
    return dict(kappa_input=kappa,fitted_kappa=float(W@signal),
        bare_K02=float(K[0,2]),relaxed_K02=float(Keff[0,1]),
        static_covariance_U02_A2=float(U[0,2]),
        maximum_C1_abs=float(np.max(np.abs(null))),
        maximum_C3_abs=float(np.max(np.abs(pred))),
        maximum_C5_abs=float(np.max(np.abs(fifth))),
        maximum_physical_mask_error_eV3=float(np.max(abs(signal-pred))),
        sigma_kappa_at_Ncounts=float(np.sqrt(1/(F@(F/variance)))),
        Ncounts=Ncounts,
        q_range_Ainverse=[float(qs[0]),float(qs[-1])],
        temperature_K=295.)

def phip_spin_energy_budget(alpha=.964,frequency_Hz=400e6,delta_Hz=492.,J_Hz=4.6):
    """Exact Zeeman/J sideband moment corrections for ideal stored pair order.

    These coefficients multiply measured lower mechanical moments. They do
    NOT bound conformation dependence J(R), reaction heating, spin transfer
    between molecules, or other material changes. The imperfect RF family
    needs its own measured spin trajectories in the same calculation.
    """
    H,D=hh_operators();hz=constants.h/constants.electron_volt
    hs=hz*(-frequency_Hz*(H[2]+D[2])+delta_Hz/2*(H[2]-D[2])+
        J_Hz*sum(H[k]@D[k] for k in range(3)))
    ens,V=np.linalg.eigh(hs);U=np.kron(V,np.eye(2));en=np.repeat(ens,2)
    pauli=[2*x for x in angular_momentum(.5)]
    bs=[U.conj().T@(-3.74*np.eye(8)+29.12*sum(np.kron(group[k],pauli[k]) for k in range(3)))@U for group in [H,D]]
    de=en[:,None]-en[None,:];allmom=[]
    for sign in [1,-1]:
        rho=np.eye(4)/4+sign*alpha*H[2]@D[2]
        assert np.max(abs(rho@hs-hs@rho))<1e-20
        pop=np.repeat(np.diag(V.conj().T@rho@V).real/2,2)
        M=np.zeros((4,2,2))
        for n in range(4):
            for i in range(2):
                for j in range(2):M[n,i,j]=np.sum(pop[None,:]*de**n*bs[i].conj()*bs[j]).real
        allmom.append(M)
    mm=np.array(allmom);diff=mm[0]-mm[1];ref=mm[:,0].mean(axis=0)
    return dict(difference_spin_moments=diff.tolist(),
        eta_r_for_lower_moment_bound={str(r):raw_reference_bound(diff[r],ref) for r in [1,2,3]},
        contrast_fm2=alpha*29.12**2,
        energy_eigenvalues_eV=ens.tolist())
```

## 7. HD pair order: exact gate, useful failure

For proton I=1/2 and deuteron S=1, start with

\[
\rho_0=\mathbf1_6/6+(P/3)I_z.
\]

This assumes an initially maximally mixed deuteron, an explicit preparation requirement. A proton π/2 pulse, evolution under 2πJ I_z S_z for t=1/(4J), and a second proton π/2 pulse produce

\[
\rho_\pm=\mathbf1_6/6+(P/3)[I_x\Pi_{S_z=0}\pm I_z S_z]
\]

up to pulse-sign conventions. A proton π pulse reverses the longitudinal correlation while preserving the common transverse term. Both one-spin marginals match; pair dot products are ±P/3. The two-molecule trace verifies that only the intramolecular H–D weights change in the product-spin model. Global phase cycling is not independent local dephasing: its residual intermolecular transverse correlations must be checked, and happen to cancel here.

## 8. Stronger physical source: parahydrogen-derived two-proton order

Anwar and colleagues demonstrated nearly pure, addressable hydride spins in Ru(H)2(CO)2(dppe) in solution. The reported spin populations were 0.937 singlet, 0.045 T0, and 0.009 each T+ and T−. A separate paper demonstrates coherent control with hydride separation 492 Hz, scalar coupling 4.6 Hz, T1=1.7 s and T2=0.58 s, using phosphorus decoupling. These are **realized spin components**, not a neutron scattering assembly.

The useful resource here is longitudinal pair order, which does not require transverse coherence throughout readout. An I/Z phase cycle on one addressed hydride maps the reported state to

\[
\rho_-={\mathbf1_4\over4}-\alpha I_zS_z,
\quad \alpha=0.937+0.045-0.009-0.009=0.964.
\]

A selective proton π pulse gives rho_+=1/4+alpha I_z S_z. Both marginals are maximally mixed. The states are separable; entanglement is not needed for the stored contrast. They commute with both the Zeeman Hamiltonian and the isotropic two-spin J coupling. For molecular spin states that remain independent between molecules,

\[
I_+-I_-=\alpha c_H^2\Re\sum_\ell S_{H_{\ell,1}H_{\ell,2}}.
\]

Same-atom and all inter-molecular contributions cancel in the static-spin limit, including mechanically correlated neighbours. No mechanical pair potential is subtracted. The computed coefficient is about 817 fm² per selected pair-spectrum normalization. **Alpha is not the paper's Werner-state approximation ε≈0.916.** The source populations, not an assumed pure singlet, set this number.

The microscopic model for this proposal is H_total=H_mechanical(R,p)+H_spin(R)+H_RF(t)+H_neutron+Σ_i (2πhbar²/m_n)b_i δ(r_n-R_i). The leading calculation treats H_spin's measured parameters as coordinate independent and the mechanical state as unchanged by the spin control. Conformation-dependent J(R), spin–rotation couplings, spin transfer between molecules and photochemical/thermal history are additional physical terms, not silently supplied operations.

### A physical selective pulse and its error

One proposed rectangular pulse uses the same RF field on both hydrides. At resonance with the first, set f_R=492/sqrt(3) Hz and duration 1/(2 f_R). Without J, the target spin rotates by π and the other by 2π. The code retains the measured J during the pulse. The required co-rotating B1 is 6.67 μT and the duration 1.760 ms. This is a specified design calculation, not a reported pulse setting. The finite pulse produces small single-spin magnetizations and unwanted intermolecular weights. A two-molecule leakage norm is not automatically a uniform bulk bound.

## 9. A calibrated physical family, including finite-pulse backreaction

For a product of identical molecular spin density matrices, every intermolecular AA, BB or AB spin weight is determined by the one-spin means mA and mB. Intramolecular scattering depends instead on Q=<IA·IB>. An experimentally calibrated row for a control setting can therefore contain

\[
(1,\overline{|m_A|^2},\overline{|m_B|^2},\overline{m_A\cdot m_B},
\overline{m_{Az}},\overline{m_{Bz}},\overline Q).
\]

The overbar is an acquisition-time average. Products must be averaged *before* averaging the means. The two linear columns cover thermal spectator polarization along the field and relevant linear spin nuisances. Unpolarized spectator terms otherwise cancel with the constant column. Correlated spectator spins require additional information.

The construction below uses 16 durations, equally spaced from 0 to 7.041 ms, at the same 284.056 Hz RF amplitude. It then propagates each state during a 20 ms hold under the measured chemical-shift difference and J. All means, pair order and finite-pulse errors follow from this one spin Hamiltonian. Finite-window time averages are evaluated analytically using exp(-iπfT)sinc(fT), including frequency sums for products of means.

Weights w satisfy A^T w=(0,0,0,0,0,0,alpha/2). Consequently all intermolecular nuclear-spin contrast cancels **for any mechanical spectrum or arrangement of the molecules** within the product-spin/uniform-control premises. This avoids a Gaussian spatial shell fit. The weights depend on spin calibration, not on the mechanical forces being tested.

This is not a guarantee that the real specimen has these rows. NMR controls must acquire the actual time-dependent spin state, RF field/age distribution and absolute pair order in the neutron-visible volume. Spatial averaging of spin populations is insufficient when molecules experience different controls. Intermolecular spin transfer invalidates factorization. The source experiment itself required a measured active-volume correction; its high-purity cohort was extremely dilute. Neither high purity nor measured T1/T2 provides a replenishment rate or a neutron-sized inventory.

The code also supplies a strict warning: an ensemble made from |up down up down> and |down up down up> has exactly the same local anti-correlated pair marginals as independently prepared anti-correlated pairs, but an unwanted inter-molecular <I1z I4z>=-1/4. Local longitudinal pair tomography alone cannot certify the absence of neighbouring spin tags. An initial independent singlet source plus a controlled/independently bounded spin-transfer channel is part of the physical construction.

## 11. Noncircular calibration and spectral-tail bounds

Let T be the desired pair matrix (T_ij=T_ji=1/2 per selected unordered pair). Suppose independent spin/beam calibration gives

\[
\sum_k w_k G_k=D T+R,
\quad D=\alpha c_H^2.
\]

Let Gref be a known positive reference matrix supplied by a convex combination of actually acquired settings. On its supported subspace, define

\[
\eta=\|G_{ref}^{-1/2} R G_{ref}^{-1/2}\|_{op}.
\]

Because the mechanical spectral matrix S(q,E) is positive,

\[
|\operatorname{Tr}(R S)|\le\eta\operatorname{Tr}(G_{ref}S)=\eta I_{ref}(q,E).
\]

Thus, over a measured window W,

\[
|\delta C_3^W|\le {\eta\over |D|}\int_W |E|^3 I_{ref}(q,E)dE.
\]

This uses a spin-weight error and a measured positive spectrum, not the unknown force field. It does not certify eta from bulk polarization alone. Matrix support, nonuniform controls, correlated spins and missing populations must be included in calibration; a small two-molecule eta cannot be extrapolated blindly to a bulk coherent mode. The finite physical-family calculation has sum|w|≈2.679, which also provides an explicit worst-case amplification for bounded calibration-column errors.

For tail T outside the measured window, positivity gives another independent certificate:

\[
\left|\int_T E^3\Re S_{AB}\,dE\right|
\le\left[\int_T |E|^3S_{AA}\,dE
\int_T |E|^3S_{BB}\,dE\right]^{1/2}.
\]

Those positive tails must themselves be measured or independently bounded. Simulating them with the force field under test does not certify the experiment. A positive raw reference can also provide a looser matrix-norm bound. A finite measured energy band and C1=0 alone never give such a tail certificate: cross lines at L, 2L, 3L with weights a,-2a,a leave both C0 and C1 unchanged but change C3 by 12aL³. Positive diagonal weights can accompany each line. Additional constraints must be stated to rule out such high-energy ambiguity.

Spin sidebands are a separate, computable correction. For a separable stationary spin/mechanical Hamiltonian, exact convolution gives

\[
M_3^{total}=\sum_{r=0}^3 {3\choose r}\sum_{ij}G_r^{ij}M_{3-r}^{ij,mechanical}.
\]

The spin matrices Gr are calculated from independent NMR energies and populations. The HD and ideal stored PHIP pair-order calculations above include Zeeman and J splittings. They do not establish that J is coordinate independent: the force effect of J(R), spin–rotation interactions and preparation bias requires its own bound. A small measured J value alone is not a rigorous derivative bound.

Finally, for independent raw Poisson measurements with exposure Lk,

\[
\operatorname{Var}(\widehat C_3)=D^{-2}\sum_k {w_k^2\over L_k}
\int E^6 I_k(q,E)dE.
\]

This is why removing self terms in the mean does not remove their statistical cost. Energy-gain data at finite temperature, background subtraction, efficiency, energy calibration, and the actual q–E response belong in these raw moments. The finite-family result 1.57 is a contrast-only equal-intensity standard-deviation ratio, not an instrument improvement factor.

## 12. What the construction adds, and what it does not

The surviving operation is a controlled pair-response measurement: independently prepared nuclear labels select a particular intramolecular pair class while atomic masses, chemistry and mechanical coordinates stay the same across spin settings. This can supply the exact mixed-curvature diagnostic without reconstructing all environmental forces. It does not select an arbitrary single molecule, write arbitrary pair masks, recover a full 3D Hessian from powder data, or prove chemical causality from a single scalar.

The strongest comparators receive the same structural and spin calibration information:

* **Coherent vibrational spectroscopy:** interference/eigenvector-sensitive x-ray methods are realized; full normal-mode energies and eigenvectors determine a harmonic Hessian. If they provide the same localized cross spectrum, they provide the same moment information. The proposed labels are an alternative coupling route, not an information monopoly.
* **Existing NMR-modulated neutron scattering:** directly relevant prior for spin-selected partial structure factors. The present construction must stand on an exact force observable, explicit nuisance elimination, and usable errors, not on claiming spin contrast itself.
* **NRIXS and neutron-Compton potential reconstruction:** realized moment/potential methods already recover force-related or anharmonic information. A joint selected-pair observable has a different target from an on-site force projection or a one-coordinate effective potential.
* **Static force reconstruction:** complete harmonic covariance/eigenmode information can recover forces; partial structural distributions generally cannot, as the explicit identifiability construction shows. The bridge's Schur-complement coupling is a meaningful relaxed response, not an erroneous measurement.
* **Direct intervention:** when coordinates can actually be displaced/clamped and forces read independently, mixed force derivatives can be measured directly. Optical-trap force measurements are an established realization at larger scales. The label method does not outrank that route by denying it access to known geometry or calibration.

A low-q limit would remove the phase weighting without losing spin pair selection, but neutron kinematics must be paid for. At finite vibrational energy, q cannot approach zero at fixed incident energy. This notebook does not grant an unattained low-q/high-energy detector or use a one-phonon approximation to dispose of tails.

**Current allocation:** retain the physical spin-contrast family and the direct-versus-relaxed discriminator. The cold dilute-HD implementation is a poor practical carrier at its demonstrated thermal polarization. The parahydrogen-derived pair-order route has a stronger acquired actuator but lacks a demonstrated combination of absolute retained pair inventory, controlled intermolecular spin leakage and certified high-energy spectral coverage. A further numerical optimum without those inputs would not establish a usable instrument. The immediate material question is whether an actual pair-order specimen can satisfy the noncircular leakage/tail/count error budget; the present calculations neither assume that answer nor turn its absence into a universal no-go theorem.
